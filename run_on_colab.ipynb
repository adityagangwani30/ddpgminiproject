## 1) Set Repo Details

`REPO_URL` is pre-filled for this project. Change only if you are using a fork.

```python
import os
import subprocess
from pathlib import Path

REPO_URL = "https://github.com/adityagangwani30/ddpgminiproject.git"
BRANCH = "main"  # <-- change if needed

repo_name = REPO_URL.rstrip("/").split("/")[-1].replace(".git", "")
repo_path = Path("/content") / repo_name

if repo_path.exists():
    print(f"Repo already exists: {repo_path}")
else:
    subprocess.run(["git", "clone", "-b", BRANCH, REPO_URL], check=True)

%cd /content/{repo_name}
!git status -sb
```

## 3) Train DDPG

Choose mode: `quick`, `balanced`, or `full`.

```python
TRAIN_MODE = "balanced"  # quick | balanced | full
TIMESTEPS = None  # e.g., 50000 for full report, or None

cmd = ["python", "train_ddpg.py", "--mode", TRAIN_MODE]
if TIMESTEPS is not None:
    cmd.extend(["--timesteps", str(TIMESTEPS)])

print("Running:", " ".join(cmd))
subprocess.run(cmd, check=True)
```
